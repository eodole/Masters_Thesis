```python
import numpy as np 
import utils 
import networkx as nx 
```

```python
true_graph = nx.DiGraph()

true_graph.add_edges_from([(1,3), (3,4), (2,4), (2,7), (4,6), (6,5)])

custom_labels = {1: 1, 2: 4, 3: 2, 4: 3, 5: 7, 6: 6, 7: 5}
nx.draw_networkx(true_graph, labels=custom_labels)
```

```python


def assign_weights(A): 
    
    r_weights = np.random.rand(A.shape[0],A.shape[1])
    
    return(np.multiply(A,r_weights))

def gen_data(weights, n_samples):
    d_feats = weights.shape[0]
    data = np.ones(shape=(n_samples, d_feats))
    noise = np.random.normal(size=(n_samples, d_feats))

    return np.add(noise, np.matmul(data, weights))

```

```python
#true graph
true_A = nx.to_numpy_array(true_graph)
true_W = assign_weights(true_A)

true_df = gen_data(true_W, 10000)

# np.savetxt("./exp5/exp5_zscore_true_data.csv", true_df, delimiter=",")
# np.savetxt("./exp5/exp5_zscore_true_adj.csv", true_A, delimiter=",")
```

```python
np.savetxt("./exp5/exp5_zscore_true_weights.csv", true_W, delimiter=",")
```
